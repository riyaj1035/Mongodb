# Theoretical Questions

1. What are the key differences between SQL and NoSQL databases
-  SQL databases store data in tables with rows and columns and use a predefined schema, whereas NoSQL databases store data in formats such as documents, key-value pairs, and graphs with flexible schemas. SQL databases are best suited for structured data and complex queries, while NoSQL databases are useful for handling large amounts of flexible or unstructured data. Examples of SQL databases are MySQL and PostgreSQL, while MongoDB and Cassandra are examples of NoSQL databases.

2. What makes MongoDB a good choice for modern applications
-  MongoDB is a good choice for modern applications because it offers a flexible schema, high scalability, fast data access, and the ability to store complex data in JSON-like documents.

3. Explain the concept of collections in MongoDB.
-  In MongoDB, a collection is a group of related documents, similar to a table in a SQL database. Collections store documents in BSON format, and each document can have a different structure or set of fields. For example, a students collection can store student details such as name, age, and marks.
-  Example: A students collection may contain documents for multiple students.

4. How does MongoDB ensure high availability using replication
- MongoDB ensures high availability through replication using replica sets, which maintain multiple copies of the same data across different servers. A replica set has a primary node that handles write operations and secondary nodes that replicate data from the primary. If the primary node fails, one of the secondary nodes is automatically elected as the new primary, ensuring the database remains available with minimal downtime.

5. What are the main benefits of MongoDB Atlas
- MongoDB Atlas is a fully managed cloud database service that offers easy deployment, automatic scaling, high availability, automated backups, strong security, and monitoring tools. It reduces the need for manual database management and allows developers to focus on building applications.**bold text**

6. What is the role of indexes in MongoDB, and how do they improve performance
- Indexes in MongoDB improve query performance by allowing the database to find and retrieve documents quickly without scanning every document in a collection. They make searching, sorting, and filtering faster, but they require additional storage space and can slow down write operations because indexes must be updated when data changes.*italicised text*

7. Describe the stages of the MongoDB aggregation pipeline.
- The MongoDB aggregation pipeline processes documents through multiple stages, such as $match to filter documents, $group to group data and perform calculations, $sort to sort results, $project to select or reshape fields, and $limit to restrict the number of documents returned. Each stage passes its output to the next stage to analyze and transform data.

8. What is sharding in MongoDB? How does it differ from replication
- Sharding in MongoDB distributes large datasets across multiple servers to improve scalability and handle large workloads, while replication maintains multiple copies of the same data across servers to ensure high availability and fault tolerance. Sharding is used for horizontal scaling, whereas replication is used for data redundancy and availability.

9. What is PyMongo, and why is it used
- PyMongo is a Python library that allows Python applications to connect to and interact with MongoDB databases. It is used to perform database operations such as inserting, retrieving, updating, and deleting documents, as well as managing collections and executing queries.

10. What are the ACID properties in the context of MongoDB transactions
- The ACID properties in MongoDB transactions are **Atomicity, Consistency, Isolation, and Durability**. Atomicity ensures that all operations in a transaction succeed or none are applied; Consistency maintains data integrity; Isolation ensures concurrent transactions do not interfere with each other; and Durability guarantees that committed changes are preserved even after a system failure.

11. What is the purpose of MongoDB’s explain() function
- MongoDB’s `explain()` function is used to analyze how a query is executed and identify ways to improve its performance. It provides information about query execution plans, index usage, and the number of documents examined and returned, helping developers optimize queries and improve database efficiency.

12. How does MongoDB handle schema validation
- MongoDB handles schema validation using validation rules, such as JSON Schema, to ensure that documents follow specified data types, required fields, and other constraints before they are inserted or updated in a collection.

13. What is the difference between a primary and a secondary node in a replica set
- In MongoDB, the **primary node** handles all write operations and records changes in the operation log, while the **secondary nodes** replicate data from the primary node and can serve read operations if configured. If the primary node fails, a secondary node can be elected as the new primary to maintain high availability.

14. What security mechanisms does MongoDB provide for data protection
- MongoDB provides security mechanisms such as authentication, role-based access control (RBAC), encryption of data at rest and in transit, auditing, and network access restrictions to protect data from unauthorized access and ensure database security.

15. Explain the concept of embedded documents and when they should be used.
- In MongoDB, **embedded documents** are documents nested inside another document, used to store related data together. They should be used when data is closely related, frequently accessed together, and does not grow excessively, as they improve query performance and reduce the need for joins.

16. What is the purpose of MongoDB’s $lookup stage in aggregation
- MongoDB’s **`$lookup`** stage is used to perform a left outer join between two collections, allowing documents from one collection to be combined with matching documents from another collection based on a specified field.

17. What are some common use cases for MongoDB
- MongoDB is commonly used for content management systems, e-commerce applications, real-time analytics, social media platforms, IoT applications, and mobile applications because it offers a flexible schema, high scalability, and efficient data storage and retrieval.

18. What are the advantages of using MongoDB for horizontal scaling?
- MongoDB supports horizontal scaling through **sharding**, which distributes data across multiple servers to handle large datasets and increasing workloads, improve performance, and provide greater storage capacity without relying on a single powerful server.

19. How do MongoDB transactions differ from SQL transactions?
- MongoDB transactions support ACID properties across multiple documents and collections, while SQL databases typically support ACID transactions across relational tables. MongoDB uses a flexible document model, whereas SQL databases use structured tables with predefined schemas. Both support commit and rollback operations to maintain data consistency.

20. What are the main differences between capped collections and regular collections?
- In MongoDB, **capped collections** have a fixed size and automatically overwrite the oldest documents when they reach their limit, while **regular collections** can grow dynamically and do not automatically remove old documents. Capped collections are useful for logs and event records, whereas regular collections are suitable for general-purpose data storage.

21. What is the purpose of the $match stage in MongoDB’s aggregation pipeline?
- The **`$match`** stage in MongoDB’s aggregation pipeline filters documents based on specified conditions, allowing only matching documents to pass to the next stage, which improves query efficiency by reducing the amount of data processed.

22. How can you secure access to a MongoDB database?
- we can secure access to a MongoDB database by enabling authentication, using role-based access control (RBAC), encrypting data in transit and at rest, restricting network access, and assigning users only the permissions they need.

23. What is MongoDB’s WiredTiger storage engine, and why is it important?
- MongoDB’s **WiredTiger storage engine** is the default storage engine that manages how data is stored, accessed, and modified. It is important because it supports document-level concurrency control, compression, and efficient memory usage, improving database performance and scalability.

1. Write a Python script to load the Superstore dataset from a CSV file into MongoDB.

```python

import pandas as pd
from pymongo import MongoClient

# Connect to MongoDB
client = MongoClient("mongodb://localhost:27017/")

# Create or access database
db = client["SuperstoreDB"]

# Create or access collection
collection = db["Orders"]

# Load CSV dataset
df = pd.read_csv("superstore.csv")

# Convert CSV data into dictionaries
data = df.to_dict("records")

# Insert records into MongoDB
if data:
    collection.insert_many(data)

print("Superstore dataset successfully loaded into MongoDB!")

# Display total records
print("Total records:", collection.count_documents({}))

# Close connection
client.close()
```

7. Update all orders with Ship Mode as "First Class" to "Premium Class"

```python

from pymongo import MongoClient

# Connect to MongoDB
client = MongoClient("mongodb://localhost:27017/")

# Access database and collection
db = client["SuperstoreDB"]
collection = db["Orders"]

# Update all matching orders
result = collection.update_many(
    {"Ship Mode": "First Class"},
    {"$set": {"Ship Mode": "Premium Class"}}
)

# Display update result
print("Matched documents:", result.matched_count)
print("Modified documents:", result.modified_count)

# Close connection
client.close()
```

9. Use aggregation to group orders by Region and calculate total sales per region.

```python

from pymongo import MongoClient
from pprint import pprint

# Connect to MongoDB
client = MongoClient("mongodb://localhost:27017/")

# Access database and collection
db = client["SuperstoreDB"]
collection = db["Orders"]

# Aggregation pipeline
pipeline = [
    {
        "$group": {
            "_id": "$Region",
            "Total_Sales": {"$sum": "$Sales"}
        }
    },
    {
        "$sort": {"Total_Sales": -1}
    }
]

# Execute aggregation
result = collection.aggregate(pipeline)

# Display results
for region in result:
    pprint(region)

# Close connection
client.close()
```

11. Count the number of orders for each category.

```python

from pymongo import MongoClient
from pprint import pprint

# Connect to MongoDB
client = MongoClient("mongodb://localhost:27017/")

# Access database and collection
db = client["SuperstoreDB"]
collection = db["Orders"]

# Aggregation pipeline
pipeline = [
    {
        "$group": {
            "_id": "$Category",
            "Total_Orders": {"$sum": 1}
        }
    },
    {
        "$sort": {"Total_Orders": -1}
    }
]

# Execute aggregation
result = collection.aggregate(pipeline)

# Display results
print("Number of orders in each category:")
for category in result:
    pprint(category)

# Close connection
client.close()
```
